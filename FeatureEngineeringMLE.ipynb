# OUTLIERS

```python
# To check outlier presence -- BOXPLOT
sns.boxplot(x=df['ApplicantIncome'])
```

```python
# IQR METHOD
Q1 = df['ApplicantIncome'].quantile(0.25)
Q3 = df['ApplicantIncome'].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print(lower_limit)
print(upper_limit)
df.loc[df['ApplicantIncome']>upper_limit]
df.loc[df['ApplicantIncome']>upper_limit,'ApplicantIncome'] = df['ApplicantIncome'].median()
df.loc[df['ApplicantIncome']>upper_limit]
```

```python
sns.boxplot(x=df['CoapplicantIncome'])
```

```python
# IQR METHOD
Q1 = df['CoapplicantIncome'].quantile(0.25)
Q3 = df['CoapplicantIncome'].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print(lower_limit)
print(upper_limit)
df.loc[df['CoapplicantIncome']>upper_limit]
df.loc[df['CoapplicantIncome']>upper_limit,'CoapplicantIncome'] = df['CoapplicantIncome'].median()
df.loc[df['CoapplicantIncome']>upper_limit]
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['0','1','2','3']])
x[['Dependents']] = oe.fit_transform(x[['Dependents']])
```

```python
x = pd.get_dummies(x,columns=['Gender','Married','Education','Self_Employed','Property_Area'],drop_first=True)
```

```python
Trans_cols = ['ApplicantIncome','CoapplicantIncome','LoanAmount','Loan_Amount_Term']
for col in Trans_cols:
    sns.histplot(x[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
# Transformation 
x['ApplicantIncome'] = np.log1p(x['ApplicantIncome']) # Highly right skewed
x['CoapplicantIncome'] = np.log1p(x['CoapplicantIncome'])  # Moderately right skewed
x['LoanAmount'] = np.log1p(x['LoanAmount'])
# Moderately skewed -- Best Transformation is Square Root OR Log
# But better to go with Log Transformation
# because loan amounts usually behave like income distribution


# Loan_Amount_Term -2.40 -- Highly negatively skewed
# But Important Observation -- most values are same (like 360 months)
# So this column is often almost constant/discte.
# Transformation usually does NOT help much here.
```

```python
import pandas as pd
from textblob import TextBlob
```

```python
df = pd.read_excel('sentiment_dataset_1000.xlsx')
```

```python
df
```

```python
print(df.info())
```

```python
df['actual_label'].value_counts()
```

```python
def pred(text):
    polarity = TextBlob(text).sentiment.polarity
    if polarity > 0:
        return 'Positive'
    elif polarity < 0:
        return 'Negative'
    else:
        return 'Neutral'

df['predicted_label'] = df['text'].apply(pred)
```

```python
df['predicted_label']
```

```python
df['polarity'] = df['text'].apply(lambda x: TextBlob(x).sentiment.polarity)
df['subjectivity'] = df['text'].apply(lambda x: TextBlob(x).sentiment.subjectivity)
df.head()
```

```python
df
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(df['actual_label'], df['predicted_label'])
print('Accuracy:', accuracy)
```

```python
df.to_excel('sentiment_analysis_results.xlsx', index=False)
```
